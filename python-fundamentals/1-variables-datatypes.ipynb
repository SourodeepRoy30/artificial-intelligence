## Variables and Data Types

### Must Follow (Syntax Rules)
1. Must start with a letter (a-z, A-Z) or an underscore (`_`) — never a digit.
   - Valid: `name`, `_value`, `x1`
   - Invalid: `1value`, `2nd_item`
2. Can only contain letters, digits, and underscores — no spaces or special characters (`-`, `@`, `!`, `%`, etc.)
   - Valid: `user_name`, `doc_id_1`
   - Invalid: `user-name`, `doc id`, `price$`
3. Case-sensitive — `Name`, `name`, and `NAME` are three different variables.
4. Cannot be a reserved keyword (e.g. `if`, `for`, `class`, `return`, `import`, `True`, `None`, `lambda`).
   - Check reserved words anytime with:
```python
     import keyword
     print(keyword.kwlist)
```
5. No length limit, but keep it reasonable and readable.

### Best Practices (Style Conventions — PEP 8)
6. Use `snake_case` for variable and function names.
   - Good: `user_name`, `retrieval_score`, `chunk_size`
   - Avoid: `UserName`, `retrievalScore` (these are conventions for classes/other languages)
7. Use `PascalCase` for class names.
   - Example: `class DocumentRetriever:`
8. Use `UPPER_SNAKE_CASE` for constants.
   - Example: `MAX_TOKENS = 4096`
9. Prefix with a single underscore (`_variable`) to indicate "internal use" by convention (not enforced by Python).
10. Use double underscore prefix (`__variable`) only when you specifically want name-mangling in classes (advanced, rarely needed early on).
11. Avoid single-character names except for short-lived loop counters (`i`, `j`, `k`) or well-known math contexts (`x`, `y`).
12. Be descriptive and specific — favor clarity over brevity.
    - Good: `embedding_vector`, `similarity_score`
    - Avoid: `e`, `s`, `data2`
13. Avoid shadowing built-in names (`list`, `dict`, `str`, `type`, `id`, `input`) as variable names — it overwrites the built-in function/type for that scope.
    - Avoid: `list = [1, 2, 3]`
    - Better: `my_list = [1, 2, 3]`

### Quick Reference Table

| Type | Convention | Example |
|---|---|---|
| Variable / function | `snake_case` | `chunk_text`, `get_embedding()` |
| Class | `PascalCase` | `class RAGPipeline:` |
| Constant | `UPPER_SNAKE_CASE` | `TOP_K = 5` |
| "Internal" / private-ish | `_leading_underscore` | `_cache` |
| Name-mangled (class-private) | `__double_leading` | `__weights` |

## Data Types with Methods

Every data type in Python is an object, so technically all have methods — 
but some types have rich, frequently-used method sets while others have very few.

### `str` — Text Processing
| Method | Purpose | Example |
|---|---|---|
| `.upper()` / `.lower()` | Change case | `"hi".upper()` → `"HI"` |
| `.strip()` | Remove leading/trailing whitespace | `"  hi  ".strip()` → `"hi"` |
| `.split()` | Break string into a list by delimiter | `"a,b,c".split(",")` → `['a','b','c']` |
| `.join()` | Combine a list into a string | `",".join(['a','b'])` → `"a,b"` |
| `.replace()` | Replace substring | `"hi".replace("h","b")` → `"bi"` |
| `.find()` | Return index of first match (-1 if not found) | `"hello".find("l")` → `2` |
| `.startswith()` / `.endswith()` | Check prefix/suffix | `"file.txt".endswith(".txt")` → `True` |
| `.format()` | Insert values into a string | `"{} is {}".format("x", 5)` |

### `list` — Ordered Mutable Collections
| Method | Purpose | Example |
|---|---|---|
| `.append()` | Add item to end | `[1,2].append(3)` → `[1,2,3]` |
| `.extend()` | Add multiple items from another iterable | `[1].extend([2,3])` → `[1,2,3]` |
| `.insert()` | Add item at a specific index | `[1,3].insert(1,2)` → `[1,2,3]` |
| `.remove()` | Remove first matching value | `[1,2,3].remove(2)` → `[1,3]` |
| `.pop()` | Remove & return item at index (default last) | `[1,2,3].pop()` → `3` |
| `.sort()` | Sort in place | `[3,1,2].sort()` → `[1,2,3]` |
| `.reverse()` | Reverse in place | `[1,2,3].reverse()` → `[3,2,1]` |
| `.index()` | Find index of first matching value | `[1,2,3].index(2)` → `1` |

### `dict` — Key-Value Mapping
| Method | Purpose | Example |
|---|---|---|
| `.keys()` | Get all keys | `{"a":1}.keys()` → `dict_keys(['a'])` |
| `.values()` | Get all values | `{"a":1}.values()` → `dict_values([1])` |
| `.items()` | Get key-value pairs | `{"a":1}.items()` → `dict_items([('a',1)])` |
| `.get()` | Safely fetch a value (no crash if key missing) | `{"a":1}.get("b")` → `None` |
| `.update()` | Merge another dict in | `{"a":1}.update({"b":2})` |
| `.pop()` | Remove & return a key's value | `{"a":1}.pop("a")` → `1` |
| `.setdefault()` | Get value, or set + return a default if missing | `{}.setdefault("a", 0)` → `0` |

### `set` — Unique, Unordered Collections
| Method | Purpose | Example |
|---|---|---|
| `.add()` | Add an element | `{1,2}.add(3)` → `{1,2,3}` |
| `.remove()` | Remove an element (errors if missing) | `{1,2}.remove(2)` → `{1}` |
| `.discard()` | Remove an element (no error if missing) | `{1,2}.discard(5)` |
| `.union()` | Combine two sets | `{1,2}.union({2,3})` → `{1,2,3}` |
| `.intersection()` | Common elements | `{1,2}.intersection({2,3})` → `{2}` |
| `.difference()` | Elements in first but not second | `{1,2}.difference({2})` → `{1}` |

### `tuple` — Immutable, Fewer Methods
| Method | Purpose | Example |
|---|---|---|
| `.count()` | Count occurrences of a value | `(1,2,2).count(2)` → `2` |
| `.index()` | Find index of first matching value | `(1,2,3).index(3)` → `2` |

### `int` / `float` — Numeric, Few but Useful

**Casting**: Converting one data type to another data type. We use int(), float(), str(), list, set When we do arithmetic operations string numbers should be first converted to int or float otherwise it will return an error. If we concatenate a number with a string, the number should be first converted to a string. We will talk about concatenation in String section.

````python
# 1. IMPLICIT TYPE CONVERSION (Python does it automatically)

num_int = 10
num_float = 2.5

result = num_int + num_float   # int automatically becomes float during operation
print(result, type(result))    # 12.5 <class 'float'>

# Python "upgrades" int to float here to avoid losing precision.
# This ONLY happens with compatible numeric types (int -> float),
# never automatically between str and int, etc.
````

````python
# 2. EXPLICIT TYPE CONVERSION (you do it manually)

# --- int() ---
print(int(3.99))        # 3       -> truncates, does NOT round
print(int("42"))        # 42      -> string of digits to int
print(int(True))        # 1       -> bool to int
print(int(False))       # 0

# int("42.5")  # This will FAIL — int() can't parse a decimal string directly
# You'd need: int(float("42.5")) -> 42
````

````python
# --- float() ---
print(float(10))         # 10.0
print(float("3.14"))     # 3.14
print(float("10"))       # 10.0   -> works even without a decimal point
print(float(True))       # 1.0
````

````python
# --- str() ---
print(str(100))          # "100"
print(str(3.14))         # "3.14"
print(str(True))         # "True"
print(str([1, 2, 3]))    # "[1, 2, 3]"  -> even lists/dicts can become strings
````

````python
# --- bool() ---
print(bool(0))           # False
print(bool(1))           # True
print(bool(""))          # False   -> empty string is "falsy"
print(bool("hello"))     # True    -> any non-empty string is "truthy"
print(bool([]))          # False   -> empty list is "falsy"
print(bool([1, 2]))      # True
````

## Python Operators

### Arithmetic Operators
| Operator | Name | Purpose | Example | Result |
|---|---|---|---|---|
| `+` | Addition | Add two values | `5 + 3` | `8` |
| `-` | Subtraction | Subtract | `5 - 3` | `2` |
| `*` | Multiplication | Multiply | `5 * 3` | `15` |
| `/` | Division | Divide (always returns float) | `5 / 2` | `2.5` |
| `%` | Modulus | Remainder after division | `5 % 2` | `1` |
| `**` | Exponentiation | Power | `5 ** 2` | `25` |
| `//` | Floor Division | Division, rounded down to nearest integer | `5 // 2` | `2` |

### Assignment Operators
| Operator | Meaning | Example | Equivalent To |
|---|---|---|---|
| `=` | Assign value | `x = 5` | — |
| `+=` | Add and reassign | `x += 3` | `x = x + 3` |
| `-=` | Subtract and reassign | `x -= 3` | `x = x - 3` |
| `*=` | Multiply and reassign | `x *= 3` | `x = x * 3` |
| `/=` | Divide and reassign | `x /= 3` | `x = x / 3` |
| `%=` | Modulus and reassign | `x %= 3` | `x = x % 3` |
| `**=` | Exponent and reassign | `x **= 3` | `x = x ** 3` |
| `//=` | Floor divide and reassign | `x //= 3` | `x = x // 3` |

### Comparison (Relational) Operators
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `5 <= 3` | `False` |

### Logical Operators (bonus — often grouped with comparison)
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `and` | True if both conditions are true | `(5 > 3) and (2 > 1)` | `True` |
| `or` | True if at least one condition is true | `(5 > 3) or (1 > 2)` | `True` |
| `not` | Inverts a boolean | `not (5 > 3)` | `False` |

````python
text = input("Enter a string:")
reversed_text = text[::-1] #string slicing technique using sequence[start:stop:step]
# [::-1] reads as "take the whole string, but walk through it backwards."
print(reversed_text) 
````
