## Control Flow in Python

Control flow determines the **order** in which code executes. By default, Python runs 
code top to bottom, one line at a time. Control flow statements let you change that — 
skip lines, repeat lines, or branch into different paths depending on conditions.

There are two main categories:
1. **Conditional statements** — run different code depending on whether a condition is True/False
2. **Loops** — repeat a block of code multiple times

### 1. Conditional Statements (`if`, `elif`, `else`)
Used to make decisions. Python checks conditions top to bottom and runs the 
**first** block whose condition is `True`, then skips the rest.

- `if` — runs a block if a condition is true
- `elif` (else if) — checks another condition, only if the previous ones were false
- `else` — catch-all, runs if none of the above conditions were true

### 2. Nested Conditionals
An `if` statement inside another `if` statement — used when a decision depends 
on more than one layered condition.

### 3. Loops
- **`for` loop** — iterates over a known sequence (list, string, range, dict, etc.) 
  a fixed number of times
- **`while` loop** — repeats *as long as* a condition remains `True`; used when you 
  don't know in advance how many iterations you'll need

### 4. Loop Control Statements
- **`break`** — immediately exits the loop entirely
- **`continue`** — skips the rest of the current iteration, jumps to the next one
- **`pass`** — does nothing; a placeholder used when Python syntax requires a 
  statement but you don't want any action yet (e.g. an empty function body 
  you'll fill in later)

### Quick Reference Table
| Statement | Purpose | Stops loop entirely? | Skips to next iteration? |
|---|---|---|---|
| `break` | Exit loop immediately | ✅ Yes | — |
| `continue` | Skip rest of current iteration | ❌ No | ✅ Yes |
| `pass` | Do nothing (placeholder) | ❌ No | ❌ No |

````python
# 1. IF / ELIF / ELSE

age = 20

if age < 13:
    print("Child")
elif age < 20:
    print("Teenager")
else:
    print("Adult")
````

````python
# 2. NESTED CONDITIONALS

username = "roy"
password = "abc123"

entered_username = "roy"
entered_password = "abc123"

if entered_username == username:
    if entered_password == password:
        print("Login successful")
    else:
        print("Incorrect password")
else:
    print("Username not found")
````

````python
# 3. FOR LOOP — iterating over a known sequence

# Over a range of numbers
for i in range(5):
    print(i)          # 0 1 2 3 4

# Over a list
skills = ["AI", "ML", "DL"]
for skill in skills:
    print(skill)

# Over a string (iterates character by character)
for char in "Roy":
    print(char)

# Over a dictionary (iterates over keys by default)
person = {"name": "Roy", "city": "Kolkata"}
for key in person:
    print(key, ":", person[key])

# Better — iterate over key-value pairs directly
for key, value in person.items():
    print(key, ":", value)
````

````python
# 4. WHILE LOOP — repeats while condition is True

count = 0

while count < 5:
    print("Count is:", count)
    count += 1     # without this, the loop would run forever (infinite loop)
````

````python
# 5. BREAK — exit the loop immediately

for num in range(1, 10):
    if num == 5:
        break
    print(num)      # prints 1, 2, 3, 4 then stops
````

````python
# 6. CONTINUE — skip this iteration, move to next

for num in range(1, 10):
    if num % 2 == 0:
        continue     # skip even numbers
    print(num)       # prints only odd numbers: 1, 3, 5, 7, 9
````

````python
# Left-aligned triangle
n = int(input("Enter height of the triangle:"))

for i in range(1, n + 1):    # i takes values 1, 2, 3, ..., n
    for j in range(1, i + 1):        # this is where your formula from before applies
        print("*", end = " ") # end="" stops print() from adding a newline
    print()      # this print() (no arguments) moves to a new line, ONCE per row
````

````python
# Inverted left-aligned triangle
n = int(input("Enter height of the triangle: "))

for i in range(1, n + 1):          # i = 1, 2, 3, 4, 5  (same as your first triangle)
    for j in range(1, (n - i + 1) + 1):   # apply the formula: stars = n - i + 1
        print("*", end=" ")
    print()
````

````python
# Right-Aligned Triangle

n = int(input("Enter height of the triangle: "))

for i in range(1, n + 1):
    # Part 1: print leading spaces
    for s in range(1, (n - i) + 1):
        print(" ", end=" ")
    # Part 2: print stars
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
````

````python
# Inverted right-aligned triangle
n = int(input("Enter height of the triangle: "))

for i in range(1, n + 1):
    # spaces — figure out the formula
    for s in range(1, (i - 1) + 1):
        print(" ", end=" ")
    # stars — figure out the formula
    for j in range(1, (n - i + 1) + 1):
        print("*", end=" ")
    print()
````

````python
# Pyramid (Centered Triangle)
n = int(input("Enter height of the triangle: "))

for i in range(1, n + 1):
    # Part 1: print leading spaces
    for s in range(1, (n - i) + 1):
        print(" ", end=" ")
    # Part 2: print stars
    for j in range(1, (2 * i - 1) + 1):
        print("*", end=" ")
    print()
````

````python
# Inverted Pyramid (Centered Triangle)
n = int(input("Enter height of the triangle: "))

for i in range(1, n + 1):
    # Part 1: print leading spaces
    for s in range(1, (i - 1) + 1):
        print(" ", end=" ")
    # Part 2: print stars
    for j in range(1, (2*(n - i) - 1) + 1):
        print("*", end=" ")
    print()
````

## Pattern Printing — Logic, Tips & Tricks

### The Core Framework
Every star/number pattern is built from the **same skeleton**:
1. An **outer loop** — controls the row number (`i`)
2. One or two **inner loops** — one for spaces (optional), one for the actual characters
3. A `print()` with no arguments after the inner loop(s) — moves to a new line, once per row

```python
for i in range(1, n + 1):
    for s in range(1, SPACES_FORMULA + 1):
        print(" ", end=" ")
    for j in range(1, CHARACTERS_FORMULA + 1):
        print("*", end=" ")
    print()
```

The entire problem reduces to finding two formulas in terms of `i` and `n`:
- **"How many spaces in row `i`?"**
- **"How many characters in row `i`?"**

### Step-by-Step Approach for Any New Pattern
1. **Draw the pattern by hand** for a small `n` (e.g. `n = 5`), labeling each row `i = 1, 2, 3, 4, 5`.
2. **Count characters per row** and write them as a sequence: e.g. `1, 2, 3, 4, 5` or `5, 4, 3, 2, 1`.
3. **Count spaces per row** (if any) the same way.
4. **Find the formula** that produces that sequence when you plug in `i`. Common building blocks:
   - Increasing 1,2,3...n → `i`
   - Decreasing n,n-1...1 → `n - i + 1`
   - Increasing from 0 → `i - 1`
   - Decreasing to 0 → `n - i`
   - Odd numbers 1,3,5,7... → `2*i - 1`
5. **Sanity-check the formula** at the boundaries: plug in `i = 1` and `i = n` and confirm they match your hand-drawn rows.
6. **Plug the formula into `range(1, FORMULA + 1)`** — this always gives exactly `FORMULA` iterations.

### Reference Table (Formulas for Common Triangles)
| Pattern | Spaces formula | Characters formula |
|---|---|---|
| Left-aligned triangle | none | `i` |
| Inverted left-aligned triangle | none | `n - i + 1` |
| Right-aligned triangle | `n - i` | `i` |
| Inverted right-aligned triangle | `i - 1` | `n - i + 1` |
| Pyramid (centered) | `n - i` | `2*i - 1` |
| Inverted pyramid | `i - 1` | `2*(n - i) + 1` |

### Key Tips & Tricks
- **`range(start, stop)` only counts upward by default.** To count downward, you MUST add a third argument: `range(start, stop, -1)`. If in doubt, keep the outer loop counting forward (`1` to `n`) and adjust the *formula* instead of reversing the loop — it's far less error-prone.
- **`range(1, k + 1)` always gives exactly `k` iterations.** This is the standard building block — once you know `k` (your formula), you rarely need anything else.
- **`print(x, end=" ")` stops the automatic newline**, letting characters print on the same line. The plain `print()` after the inner loop(s) is what actually moves to the next row — make sure it's indented to align with the OUTER loop, not the inner one.
- **Spaces and characters are two independent formulas.** Don't try to merge them into one loop — use two separate inner loops (or two `range()` calls back to back) within the same outer iteration.
- **A "combination" pattern (e.g. diamond) is usually just two triangles printed one after another** — figure out each half separately using this same framework, then stack them.
- **Number patterns** (`1 / 1 2 / 1 2 3`) use the exact same skeleton — just replace `print("*", ...)` with `print(j, ...)`, since `j` from `range(1, k+1)` already gives you the numbers `1` through `k`.
- **Always trace your formula by hand for `i = 1` and `i = n`** before running the code — this catches 90% of off-by-one errors before you even hit run.
